# Corrected Heston MFHT validation
Only manifest-verified results are shown. Full-data refits are distinct from held-out evaluations.


```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if not (ROOT / "stabilvol").is_dir():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from stabilvol.heston.workflow import load_verified_comparison

RUN = ROOT / "data/processed/heston_corrected/ready_v2"
for market in ("UN", "LN"):
    directory = RUN / market
    for relative in ("baseline/heldout", "full_refit/export"):
        export = directory / relative
        if not (export / "manifest.json").exists():
            print(market, relative, "not yet available")
            continue
        frame = load_verified_comparison(export, directory, ROOT)
        x = (frame.lower + frame.upper) / 2
        fig, ax = plt.subplots(figsize=(9, 4))
        ax.plot(x, frame.mfht, label="empirical")
        ax.fill_between(x, frame.ci_lower, frame.ci_upper, alpha=0.2)
        ax.plot(x, frame.synthetic_mfht, label="synthetic")
        ax.fill_between(x, frame.simulation_lower, frame.simulation_upper, alpha=0.2)
        ax.set(title=f"{market}: {relative}", xlabel="Physical event volatility", ylabel="MFHT (observations)")
        ax.legend()
        ax.grid(alpha=0.2)
        plt.show()
```
